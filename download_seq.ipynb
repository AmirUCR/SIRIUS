```python
from Bio import Entrez, SeqIO
from Bio.Seq import Seq
from Bio.SeqRecord import SeqRecord

Entrez.email = "YOUR_EMAIL_HERE"  # Required by NCBI

def get_proteins(gene_id):
    # Link gene -> protein
    handle = Entrez.elink(dbfrom="gene", db="protein", id=gene_id)
    record = Entrez.read(handle)
    protein_ids = [link["Id"] for link in record[0]["LinkSetDb"][0]["Link"]]

    # Fetch all protein sequences
    handle = Entrez.efetch(db="protein", id=",".join(protein_ids), rettype="fasta", retmode="text")
    proteins = list(SeqIO.parse(handle, "fasta"))

    return proteins

IFNA2_id = "3440"
IFNA2 = ""

proteins = get_proteins(IFNA2_id)

for p in proteins:
    if p.id == "NP_000596.2":
        print(p.id, len(p.seq))
        print(p.seq)
        IFNA2 = p.seq
```

```python
url = "https://www.fpbase.org/api/proteins/?name__iexact=mcitrine&default_state__qy__gte=0.7&format=csv"

import pandas as pd

df = pd.read_csv(url)

mCitrine = df["seq"][0]

print(mCitrine)
```

```python
IFNA2_rec = SeqRecord(
    IFNA2,
    id="NP_000596.2",
    name="IFNA2",
    description="Interferon alpha 2 (human)"
)

CSF3_rec = SeqRecord(
    CSF3,
    id="NP_757373.1",
    name="CSF3",
    description="Colony stimulating factor 3 (human)"
)

EPO_rec = SeqRecord(
    EPO,
    id="NP_000790.2",
    name="EPO",
    description="Erythropoietin (human)"
)

PLAT_rec = SeqRecord(
    PLAT,
    id="NP_000921.1",
    name="PLAT",
    description="Plasminogen activator, tissue type (human)"
)

IGF1_rec = SeqRecord(
    IGF1,
    id="NP_000609.1",
    name="IGF1",
    description="Insulin like growth factor 1 (human)"
)

calb_precursor_rec = SeqRecord(
    calb_precursor,
    id="CAA83122.1",
    name="CALB",
    description="lipase B [Moesziomyces antarcticus]"
)

mcitrine_rec = SeqRecord(
    Seq(mCitrine),
    id="mcitrine",
    name="mcitrine",
    description="Modified yellow fluorescent protein published in 2002, derived from Aequorea victoria"
)

records = [IFNA2_rec, CSF3_rec, EPO_rec, PLAT_rec, IGF1_rec, calb_precursor_rec, mcitrine_rec]

output_filename = "inputs.fasta"

with open(output_filename, "w") as handle:
    SeqIO.write(records, handle, "fasta")
```
